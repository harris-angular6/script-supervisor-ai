```python
import json
from pathlib import Path

path = Path("datasets/issues/train_issues.jsonl")

with path.open("r", encoding="utf-8") as f:
    for i, line in enumerate(f):
        row = json.loads(line)
        print(json.dumps(row, indent=2, ensure_ascii=False))
        print("-" * 80)
        if i >= 4:
            break
```
