```python
import json
from pathlib import Path

path = Path("datasets/raw/train.jsonl")

shown = 0
with path.open("r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)

        for field in ["error_annotations", "annotations", "labels", "issues"]:
            if field in row and row[field]:
                print(f"\nFIELD: {field}")
                print(json.dumps({
                    "id": row.get("id"),
                    "script_id": row.get("script_id"),
                    field: row[field]
                }, indent=2, ensure_ascii=False))
                print("-" * 100)
                shown += 1
                break

        if shown >= 3:
            break
```
