```python
import json
from pathlib import Path

target = "synthetic_script_0213_error"
path = Path("datasets/scenes_enriched_error/train_error_scenes_enriched.jsonl")

count = 0
with path.open("r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)
        if row.get("script_id") == target:
            print(row["scene_id"], row["scene_index"])
            count += 1

print("count =", count)
```
