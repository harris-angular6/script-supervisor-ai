```python
import json
import random
import re
from dataclasses import dataclass
from pathlib import Path
from typing import List, Dict, Any, Tuple

# ----------------------------
# Config
# ----------------------------
SEED = 42
random.seed(SEED)

TARGET_CASES = 550            # generate 500+ (we'll over-generate a bit)
MIN_EXCERPT_CHARS = 800       # excerpt size for a "case"
MAX_EXCERPT_CHARS = 2200
MAX_CASES_PER_SCRIPT = 2      # avoid hammering one script too much

# Labels (you can expand)
LABELS = [
    "PROP_CONTINUITY",
    "WARDROBE_CONTINUITY",
    "TIME_CONTINUITY",
    "LOCATION_CONTINUITY",
    "CHARACTER_NAME_INCONSISTENCY",
    "DIALOGUE_CONTRADICTION",
    "STAGE_DIRECTION_CONTRADICTION",
]

# Safe prop/wardrobe tokens to use
PROPS = ["coffee cup", "notebook", "phone", "keys", "umbrella", "backpack", "shopping bag", "headphones", "scarf"]
WARDROBE = ["jacket", "hoodie", "coat", "hat", "glasses", "scarf", "tie", "sneakers", "boots"]
WEATHER = ["sunny", "raining", "foggy", "windy", "snowing"]
TIMES = ["MORNING", "AFTERNOON", "EVENING", "NIGHT"]

# ----------------------------
# Helpers: loading
# ----------------------------
def load_clean_texts(clean_jsonl_path: str) -> List[str]:
    """
    Expects JSONL lines like: {"id": "...", "source": "...", "text": "..."}
    """
    texts = []
    with open(clean_jsonl_path, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            obj = json.loads(line)
            t = obj.get("text", "")
            if isinstance(t, str) and len(t.strip()) > 500:
                texts.append(t.strip())
    return texts

def chunk_excerpt(text: str) -> str | None:
    """
    Extract a roughly scene-sized excerpt.
    Tries to split on scene headers like 'INT.'/'EXT.' but falls back to a random window.
    """
    text = text.replace("\r\n", "\n")

    # Find candidate scene header indices
    scene_pattern = re.compile(r"(?m)^(INT\.|EXT\.|INT\/EXT\.)[^\n]{0,80}$")
    matches = list(scene_pattern.finditer(text))
    if len(matches) >= 2:
        # choose a random scene start and take until next few headers
        i = random.randint(0, len(matches) - 2)
        start = matches[i].start()
        # end at next header or within a window
        end = matches[i + 1].start()
        excerpt = text[start:end].strip()
        if MIN_EXCERPT_CHARS <= len(excerpt) <= MAX_EXCERPT_CHARS:
            return excerpt
        # If too short, try extending to next header
        if i + 2 < len(matches):
            end2 = matches[i + 2].start()
            excerpt2 = text[start:end2].strip()
            if MIN_EXCERPT_CHARS <= len(excerpt2) <= MAX_EXCERPT_CHARS:
                return excerpt2

    # fallback: random window
    if len(text) < MIN_EXCERPT_CHARS:
        return None
    start = random.randint(0, max(0, len(text) - MAX_EXCERPT_CHARS))
    window = text[start:start + random.randint(MIN_EXCERPT_CHARS, MAX_EXCERPT_CHARS)]
    return window.strip() if len(window.strip()) >= MIN_EXCERPT_CHARS else None

# ----------------------------
# Error injectors
# ----------------------------
@dataclass
class ErrorCase:
    clean_excerpt: str
    corrupted_excerpt: str
    labels: List[str]
    explanation: str
    suggested_fix: str
    meta: Dict[str, Any]

def _replace_first(text: str, pattern: str, repl: str) -> Tuple[str, bool]:
    m = re.search(pattern, text, flags=re.MULTILINE)
    if not m:
        return text, False
    return text[:m.start()] + re.sub(pattern, repl, text[m.start():], count=1, flags=re.MULTILINE), True

def inject_prop_continuity(excerpt: str) -> ErrorCase | None:
    """
    Adds a prop in one beat then contradicts it later (has/doesn't have).
    """
    prop = random.choice(PROPS)

    # Insert a stage direction mentioning the prop after first scene header / early line
    lines = excerpt.split("\n")
    if len(lines) < 8:
        return None

    insert_idx = min(len(lines) - 1, random.randint(2, 6))
    lines2 = lines[:]
    lines2.insert(insert_idx, f"(They carry a {prop}.)")

    corrupted = "\n".join(lines2)

    # Later contradict: "empty-handed" / "no prop"
    contradiction = random.choice([
        f"(They are empty-handed now — no {prop}.)",
        f"(No {prop} is visible.)",
        f"(They set down nothing — there is no {prop}.)"
    ])
    insert2_idx = random.randint(max(insert_idx + 3, 7), min(len(lines2) - 1, insert_idx + 25))
    lines3 = corrupted.split("\n")
    lines3.insert(insert2_idx, contradiction)
    corrupted2 = "\n".join(lines3)

    explanation = f"Prop continuity issue: the excerpt introduces a {prop}, then later indicates there is no {prop} without a beat explaining the change."
    suggested_fix = f"Add a beat showing the {prop} being put away/handed off/left on a surface, or remove one of the conflicting stage directions."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted2,
        labels=["PROP_CONTINUITY"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"prop": prop}
    )

def inject_wardrobe_continuity(excerpt: str) -> ErrorCase | None:
    item = random.choice(WARDROBE)

    lines = excerpt.split("\n")
    if len(lines) < 10:
        return None

    # Mention wardrobe early
    insert_idx = random.randint(2, 7)
    lines2 = lines[:]
    lines2.insert(insert_idx, f"(They wear a {item}.)")

    # Contradict later
    insert2_idx = random.randint(min(insert_idx + 6, len(lines2) - 1), min(insert_idx + 30, len(lines2) - 1))
    lines2.insert(insert2_idx, f"(They are suddenly not wearing a {item}.)")

    corrupted = "\n".join(lines2)

    explanation = f"Wardrobe continuity issue: the character is described as wearing a {item}, then later not wearing it without any transition."
    suggested_fix = f"Add an action beat (removes/puts on the {item}) or keep wardrobe consistent by removing the conflicting description."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted,
        labels=["WARDROBE_CONTINUITY"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"wardrobe_item": item}
    )

def inject_time_continuity(excerpt: str) -> ErrorCase | None:
    """
    Flip a scene header time-of-day in one place and contradict in another nearby header.
    Works best if excerpt contains multiple scene headings.
    """
    scene_pattern = re.compile(r"(?m)^(INT\.|EXT\.|INT\/EXT\.)[^\n]*-\s*(MORNING|AFTERNOON|EVENING|NIGHT)\s*$")
    matches = list(scene_pattern.finditer(excerpt))
    if len(matches) < 1:
        return None

    corrupted = excerpt
    # Change one header time
    m = random.choice(matches)
    old_time = m.group(2)
    new_time = random.choice([t for t in TIMES if t != old_time])

    # Replace only that match instance
    header_line = m.group(0)
    new_header_line = header_line.rsplit("-", 1)[0].rstrip() + f"- {new_time}"
    corrupted = corrupted.replace(header_line, new_header_line, 1)

    # Add contradiction line later (explicit)
    contradiction = random.choice([
        f"(It still looks like {old_time.lower()}, despite the scene header.)",
        f"(The light suggests {old_time.lower()}, not {new_time.lower()}.)",
        f"(A character checks their watch: '{old_time.title()} already?')"
    ])
    lines = corrupted.split("\n")
    if len(lines) < 12:
        return None
    ins = random.randint(8, min(40, len(lines) - 1))
    lines.insert(ins, contradiction)
    corrupted2 = "\n".join(lines)

    explanation = f"Time-of-day continuity issue: a scene header was changed from {old_time} to {new_time}, but the action/dialogue still implies {old_time}."
    suggested_fix = f"Align the header and the content: either keep the header as {old_time} or adjust action/dialogue cues to match {new_time}."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted2,
        labels=["TIME_CONTINUITY"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"old_time": old_time, "new_time": new_time}
    )

def inject_location_continuity(excerpt: str) -> ErrorCase | None:
    """
    Replace INT./EXT. or location word and contradict with a stage direction.
    """
    # Try flipping INT. <-> EXT. in first scene header
    header_re = re.compile(r"(?m)^(INT\.|EXT\.|INT\/EXT\.)")
    m = header_re.search(excerpt)
    if not m:
        return None

    old = m.group(1)
    if old == "INT.":
        new = "EXT."
    elif old == "EXT.":
        new = "INT."
    else:
        new = random.choice(["INT.", "EXT."])

    corrupted = excerpt[:m.start(1)] + new + excerpt[m.end(1):]

    contradiction = random.choice([
        "(We clearly see the ceiling and enclosed walls.)",
        "(Open sky above them.)",
        "(The room echoes like an indoor space.)",
        "(Wind moves through the open air.)"
    ])
    lines = corrupted.split("\n")
    if len(lines) < 10:
        return None
    lines.insert(random.randint(6, min(25, len(lines) - 1)), contradiction)
    corrupted2 = "\n".join(lines)

    explanation = f"Location continuity issue: the scene header was changed from {old} to {new}, but the stage direction implies the opposite environment."
    suggested_fix = "Make the scene header match the described environment, or adjust stage directions to fit the intended setting."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted2,
        labels=["LOCATION_CONTINUITY"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"old_header": old, "new_header": new}
    )

def inject_character_name_inconsistency(excerpt: str) -> ErrorCase | None:
    """
    Find an ALL-CAPS character cue and introduce a near-duplicate typo later.
    """
    cues = re.findall(r"(?m)^[A-Z][A-Z0-9 ]{2,25}$", excerpt)
    cues = [c.strip() for c in cues if c.strip() and "INT." not in c and "EXT." not in c]
    cues = list(dict.fromkeys(cues))  # unique preserving order
    if not cues:
        return None

    name = random.choice(cues)
    # create an inconsistent variant (drop a letter or add space)
    if len(name.replace(" ", "")) < 4:
        return None

    variant = name
    # simple perturbation
    if " " in name and random.random() < 0.5:
        variant = name.replace(" ", "")  # e.g., "MARY JANE" -> "MARYJANE"
    else:
        # drop one character
        idx = random.randint(1, len(name) - 2)
        variant = (name[:idx] + name[idx+1:]).strip()

    # Insert a conflicting cue later
    lines = excerpt.split("\n")
    if len(lines) < 12:
        return None

    insert_at = random.randint(8, min(40, len(lines) - 1))
    lines2 = lines[:]
    lines2.insert(insert_at, variant)
    lines2.insert(insert_at + 1, "(beat)")
    corrupted = "\n".join(lines2)

    explanation = f"Character name inconsistency: the cue '{name}' appears, but later a mismatched cue '{variant}' is introduced, likely referring to the same character."
    suggested_fix = f"Standardize character cues to a single spelling ('{name}') throughout the script."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted,
        labels=["CHARACTER_NAME_INCONSISTENCY"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"name": name, "variant": variant}
    )

def inject_dialogue_contradiction(excerpt: str) -> ErrorCase | None:
    """
    Insert a pair of contradictory lines about a simple fact (weather/possession).
    """
    weather = random.choice(WEATHER)
    opposite = random.choice([w for w in WEATHER if w != weather])

    lines = excerpt.split("\n")
    if len(lines) < 12:
        return None

    insert_at = random.randint(6, min(25, len(lines) - 1))
    block = [
        "ALEX",
        f"It’s {weather} today.",
        "",
        "ALEX",
        f"No, wait — it’s actually {opposite}.",
    ]
    lines2 = lines[:insert_at] + block + [""] + lines[insert_at:]
    corrupted = "\n".join(lines2)

    explanation = f"Dialogue contradiction: the character states it’s {weather}, then immediately claims it’s {opposite} without a reason (joke, correction, or reveal)."
    suggested_fix = "Either keep one statement, or add a clarifying beat (e.g., they moved locations, they were mistaken, or it’s intentionally comedic)."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted,
        labels=["DIALOGUE_CONTRADICTION"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"weather": weather, "opposite": opposite}
    )

def inject_stage_direction_contradiction(excerpt: str) -> ErrorCase | None:
    """
    Insert conflicting stage directions: sits/stands, enters/exits, etc.
    """
    pairs = [
        ("(They sit down.)", "(They remain standing.)"),
        ("(They enter.)", "(They are already inside.)"),
        ("(They pick up the phone.)", "(They do not have a phone.)"),
        ("(They open the door.)", "(The door stays closed.)"),
        ("(They turn off the lights.)", "(The lights are still on.)"),
    ]
    a, b = random.choice(pairs)

    lines = excerpt.split("\n")
    if len(lines) < 10:
        return None

    i1 = random.randint(4, min(18, len(lines) - 1))
    i2 = random.randint(i1 + 3, min(i1 + 25, len(lines) - 1))
    lines2 = lines[:]
    lines2.insert(i1, a)
    lines2.insert(i2, b)
    corrupted = "\n".join(lines2)

    explanation = "Stage direction contradiction: the excerpt contains two action beats that cannot both be true without an intervening transition."
    suggested_fix = "Add a transition beat that resolves the change (e.g., sits then stands), or remove one of the conflicting directions."

    return ErrorCase(
        clean_excerpt=excerpt,
        corrupted_excerpt=corrupted,
        labels=["STAGE_DIRECTION_CONTRADICTION"],
        explanation=explanation,
        suggested_fix=suggested_fix,
        meta={"beat_a": a, "beat_b": b}
    )

INJECTORS = [
    inject_prop_continuity,
    inject_wardrobe_continuity,
    inject_time_continuity,
    inject_location_continuity,
    inject_character_name_inconsistency,
    inject_dialogue_contradiction,
    inject_stage_direction_contradiction,
]

# ----------------------------
# Generation loop
# ----------------------------
def generate_cases(clean_texts: List[str], target_cases: int) -> List[Dict[str, Any]]:
    cases: List[Dict[str, Any]] = []
    per_script_count = [0] * len(clean_texts)

    # safety: avoid infinite loops
    attempts = 0
    max_attempts = target_cases * 50

    while len(cases) < target_cases and attempts < max_attempts:
        attempts += 1
        idx = random.randrange(len(clean_texts))
        if per_script_count[idx] >= MAX_CASES_PER_SCRIPT:
            continue

        excerpt = chunk_excerpt(clean_texts[idx])
        if not excerpt:
            continue

        injector = random.choice(INJECTORS)
        ec = injector(excerpt)
        if ec is None:
            continue

        per_script_count[idx] += 1

        cases.append({
            "id": f"case_{len(cases):06d}",
            "source_script_index": idx,
            "labels": ec.labels,
            "clean_excerpt": ec.clean_excerpt,
            "corrupted_excerpt": ec.corrupted_excerpt,
            "explanation": ec.explanation,
            "suggested_fix": ec.suggested_fix,
            "meta": ec.meta,
        })

    if len(cases) < target_cases:
        print(f"Warning: Only generated {len(cases)} cases (target {target_cases}). "
              f"Try lowering MIN_EXCERPT_CHARS or adding more clean scripts.")
    return cases

def write_jsonl(path: str, rows: List[Dict[str, Any]]):
    with open(path, "w", encoding="utf-8") as f:
        for r in rows:
            f.write(json.dumps(r, ensure_ascii=False) + "\n")


CLEAN_JSONL = "clean_scripts.jsonl"

clean_texts = load_clean_texts(CLEAN_JSONL)
print(f"Loaded clean scripts: {len(clean_texts)}")

labeled_cases = generate_cases(clean_texts, TARGET_CASES)
print(f"Generated labeled error cases: {len(labeled_cases)}")

write_jsonl("labeled_error_cases.jsonl", labeled_cases)

# ----------------------------
# Main
# ----------------------------
'''
if __name__ == "__main__":
    # Your combined clean dataset (from previous step), JSONL with {"text": "..."} lines.
    CLEAN_JSONL = "clean_scripts.jsonl"  # <-- change

    clean_texts = load_clean_texts(CLEAN_JSONL)
    print(f"Loaded clean scripts: {len(clean_texts)}")

    labeled_cases = generate_cases(clean_texts, TARGET_CASES)
    print(f"Generated labeled error cases: {len(labeled_cases)}")

    out_path = "labeled_error_cases.jsonl"
    write_jsonl(out_path, labeled_cases)
    print(f"Wrote: {out_path}")

'''
```

```python
import json
from pathlib import Path

IN_PATH = "./labeled_error_cases.jsonl"     # change if needed
OUT_PATH = "./data/lora_sft_errors.jsonl"

SYSTEM_INSTRUCTION = (
    "You are a Script Supervisor AI assistant.\n"
    "Task: Detect continuity/consistency problems in the excerpt.\n"
    "Return ONLY valid JSON with keys:\n"
    "  labels: list of strings\n"
    "  explanation: short explanation\n"
    "  suggested_fix: short fix\n"
)

def make_prompt(corrupted_excerpt: str) -> str:
    return (
        SYSTEM_INSTRUCTION
        + "\n--- CORRUPTED EXCERPT START ---\n"
        + corrupted_excerpt.strip()
        + "\n--- CORRUPTED EXCERPT END ---\n"
        + "\nNow output the JSON."
    )

def make_completion(labels, explanation, suggested_fix) -> str:
    obj = {
        "labels": labels if isinstance(labels, list) else [],
        "explanation": explanation if isinstance(explanation, str) else "",
        "suggested_fix": suggested_fix if isinstance(suggested_fix, str) else "",
    }
    return json.dumps(obj, ensure_ascii=False)

n = 0
with open(IN_PATH, "r", encoding="utf-8") as fin, open(OUT_PATH, "w", encoding="utf-8") as fout:
    for line in fin:
        line = line.strip()
        if not line:
            continue
        ex = json.loads(line)

        corrupted = ex.get("corrupted_excerpt")
        if not isinstance(corrupted, str) or not corrupted.strip():
            continue

        row = {
            "id": ex.get("id", f"ex_{n}"),
            "prompt": make_prompt(corrupted),
            "completion": make_completion(
                ex.get("labels", []),
                ex.get("explanation", ""),
                ex.get("suggested_fix", "")
            )
        }
        fout.write(json.dumps(row, ensure_ascii=False) + "\n")
        n += 1

print("Wrote", n, "rows to", OUT_PATH)
```

```python
import json
from pathlib import Path

IN_PATH = "data/labeled_error_cases.jsonl"     # your file
OUT_PATH = "data/errors_chat.jsonl"

SYSTEM = (
    "You are a Script Supervisor AI assistant. "
    "Detect continuity/consistency problems. "
    "Reply ONLY with valid JSON with keys: labels, explanation, suggested_fix."
)

def to_chat_row(ex):
    corrupted = ex["corrupted_excerpt"].strip()
    assistant_obj = {
        "labels": ex.get("labels", []),
        "explanation": ex.get("explanation", ""),
        "suggested_fix": ex.get("suggested_fix", "")
    }
    return {
        "messages": [
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": corrupted},
            {"role": "assistant", "content": json.dumps(assistant_obj, ensure_ascii=False)}
        ]
    }

n = 0
with open(IN_PATH, "r", encoding="utf-8") as fin, open(OUT_PATH, "w", encoding="utf-8") as fout:
    for line in fin:
        line = line.strip()
        if not line:
            continue
        ex = json.loads(line)
        if "corrupted_excerpt" not in ex:
            continue
        row = to_chat_row(ex)
        fout.write(json.dumps(row, ensure_ascii=False) + "\n")
        n += 1

print("Wrote", n, "rows to", OUT_PATH)
print("Exists:", Path(OUT_PATH).exists())
```
