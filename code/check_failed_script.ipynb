```python
#synthetic_script_0564_error

import json
from pathlib import Path

path = Path("datasets/scenes_enriched/train_scenes_enriched.jsonl")  # or your current file

#target = "synthetic_script_0213_error"
target = "synthetic_script_0564_error"
found = []

with path.open("r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)
        if row.get("script_id") == target:
            found.append({
                "script_id": row.get("script_id"),
                "scene_id": row.get("scene_id"),
                "scene_index": row.get("scene_index"),
            })

print(found[:10])
print("count =", len(found))
```
